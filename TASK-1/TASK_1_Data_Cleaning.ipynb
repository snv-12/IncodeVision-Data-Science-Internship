```python
df= pd.read_csv("crime_incidents_messy.csv")
print("Dataset loaded successfully!")
df_original = df.copy()
```

```python
print("Missing Values in each column:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Unique Values in each column:")
for column in df.columns:
    print(f"{column}: {df[column].nunique()}")
```

```python
missing_values = ["?", "-", "NA", "N/A", "na", "n/a", "", "null", "NULL"]
df= df.replace(missing_values, np.nan)

print("Missing Values after Standarization:")
print(df.isnull().sum())
```

```python
duplicate_before = df.duplicated().sum()
print("Duplicate rows before cleaning:",duplicate_before)

df= df.drop_duplicates()

duplicate_after = df.duplicated().sum()
print("Duplicate rows after cleaning:",duplicate_after)
```

```python
for column in df.select_dtypes(include="object"):
    df[column] = df[column].str.strip()
```

```python
columns = ["crime_type", "district", "city", "state"]

for column in columns:
    df[column] = df[column].str.title()
```

```python
df["incident_datetime"] = pd.to_datetime(
    df["incident_datetime"],
    errors="coerce"
)
```

```python
numeric_columns = [
    "suspect_age",
    "victim_age",
    "property_loss_usd",
    "num_arrests"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
# Fill missing numerical values with median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing text/categorical values with mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])
```

```python
df["incident_datetime"] = df["incident_datetime"].fillna(
    df["incident_datetime"].mode()[0]
)
```

```python
#FINAL CHECK
print("Missing values:")
print(df.isnull().sum())
print("\nTotal Missing Values:" ,df.isnull().sum().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)
```
